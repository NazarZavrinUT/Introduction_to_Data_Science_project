**c. First shots at visualizing the data**

Take a look at the data as plots to get more insight to the patterns, trends, and relationships. This could be histograms, scatter plots, box plots, ect. What did you gather? What stands out?

```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
plt.style.use("ggplot")
```

```python
print(data.columns.to_list())
fig, axes = plt.subplots(3, 1, figsize=(17, 15))
# region_from, region_to -> mean price
pd.DataFrame({
    'region_from': data.groupby('region_from')['price'].mean(),
    'region_to':   data.groupby('region_to')['price'].mean(),
}).sort_values('region_from').plot.bar(ax=axes[0], color=['blue', 'green'], rot=0,
    xlabel='Region', ylabel='Average price', title='Average price by region')
# days_to_departure -> mean price
days_to_departure_price = data.groupby('days_to_departure')['price'].mean()
days_to_departure_price.plot(ax=axes[1], color='blue', grid=True,
    xticks=np.arange(0, days_to_departure_price.index.max() + 10, 5), rot=45,
    xlabel='Days to departure', ylabel='Average price', title='Average price by days to departure')
# travel_time -> mean price
travel_time_price = data.groupby('travel_time')['price'].mean()
travel_time_price.plot(ax=axes[2], color='blue', grid=True,
    xticks=np.arange(int(np.floor(travel_time_price.index.min() / 20) * 20), 
                    int(np.ceil(travel_time_price.index.max() / 20) * 20) + 1, 
                    20), rot=45, 
                    xlabel='Travel time', ylabel='Average price', title='Average price by travel time')
print(travel_time_price.max())
fig.tight_layout()
plt.show()
```

**Answer:** 
I gathered average price by regions of origin and destination, days to departure and travel time.

Relationships between mean prices for flying to and from a certain region differ by region.

As departure day approaches average flight price roughly increases.

Average price by travel time grows slowly until travel_time=380 minutes, then grows more quickly until travel_time=580 minutes and after that changes unevenly.
